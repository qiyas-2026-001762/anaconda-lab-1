```python
#task 2
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

movies_per_year = movies["year"].value_counts().sort_index()
print(movies_per_year)
```

```python
plt.figure(figsize=(10,5))

movies_per_year.plot(kind="line", marker="o")

plt.title("Number of Movies Released Each Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.grid(True)

plt.show()
```

```python
# Which year has the most movies
movies_per_year.idxmax()
```

```python
#task 3
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

genre_counts = movies["genre"].value_counts().head(10)
print(genre_counts)
```

```python
plt.figure

genre_counts.plot(kind="bar", color="skyblue")

plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)

plt.show()
```

```python
# Which genre contains the most movies
genre_counts.idxmax()
```

```python
# Which genre contains the fewest movies among the top 10
genre_counts.idxmin()
```

```python
#task 4 Find the top 10 directors with the largest number of movies.
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

director_counts = movies["director"].value_counts().head(10)
print(director_counts)
```

```python
#create a horizontal bar chart.
plt.figure

director_counts.sort_values().plot(kind="barh", color="orange")

plt.title("Top 10 Directors by Number of Movies")
plt.xlabel("Number of Movies")
plt.ylabel("Director")

plt.show()
```

```python
#Which director produced the most movies?
director_counts.idxmax()
```

```python
#Why is a horizontal bar chart more suitable than a vertical bar chart for this visualization
```

```python
#Task 5: Distribution of Movie Ratings Using the scorecolumn, create a histogram.
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

plt.figure(figsize=(10,5))

plt.hist(movies["score"], bins=10, color="purple", edgecolor="black")

plt.title("Distribution of Movie Ratings")
plt.xlabel("Score")
plt.ylabel("Number of Movies")

plt.show()
```

```python
#task 6: Average Movie Score by Genre Calculate the average score for each genre.
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

genre_avg_score = movies.groupby("genre")["score"].mean().sort_values(ascending=False)

print(genre_avg_score)
```

```python
#Create a bar chart.
plt.figure

genre_avg_score .plot(kind="bar", color="skyblue")

plt.title("average score by Genres")
plt.xlabel("Genre")
plt.ylabel("average score")
plt.xticks

plt.show()
```

```python
#1. Which genre has the highest average score? 
genre_avg_score .idxmax()
```

```python
#2. Which genre has the lowest average score?
genre_avg_score .idxmin()
```

```python
#Task 7: Top 10 Highest Grossing Movies Find the top 10 movies with the highest gross revenue.
import pandas as pd
import matplotlib.pyplot as plt

movies = pd.read_csv("movies_updated.csv")

gross_counts = movies["gross"].value_counts().head(10)
print(gross_counts)
```

```python
#Create a horizontal bar chart.
plt.figure

gross_counts .plot(kind="bar", color="skyblue")

plt.title("gross counts")
plt.xlabel("Gross")
plt.ylabel("score")
plt.xticks

plt.show()
```

```python
#1. Which movie earned the highest revenue?
gross_counts .idxmax()
```
