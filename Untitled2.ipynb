```python
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

standard = StandardScaler()
minmax = MinMaxScaler()
```

```python
models = {
    "SVR": SVR(),
    "Random Forest": RandomForestRegressor(random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(random_state=42),
    "Lasso": Lasso(random_state=42)
}
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

results = []

datasets = {
    "Label Encoding": X_label,
    "One-Hot Encoding": X_onehot
}

for encoding_name, data in datasets.items():

    scaling_methods = {
        "No Scaling": data,
        "StandardScaler": pd.DataFrame(
            standard.fit_transform(data),
            columns=data.columns
        ),
        "MinMaxScaler": pd.DataFrame(
            minmax.fit_transform(data),
            columns=data.columns
        )
    }

    for scaling_name, X_scaled in scaling_methods.items():

        X_train, X_test, y_train, y_test = train_test_split(
            X_scaled,
            y,
            test_size=0.2,
            random_state=42
        )

        for model_name, model in models.items():

            model.fit(X_train, y_train)

            pred = model.predict(X_test)

            score = r2_score(y_test, pred)

            results.append([
                encoding_name,
                scaling_name,
                model_name,
                score
            ])
```

```python
results_df = pd.DataFrame(
    results,
    columns=["Encoding", "Scaling", "Model", "R2 Score"]
)

results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
)

print(results_df)
```

```python
print("Best Technique")
print(results_df.iloc[0])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))

results_df_sorted = results_df.sort_values("R2 Score", ascending=False)

plt.bar(range(len(results_df_sorted)), results_df_sorted["R2 Score"], color="steelblue")

plt.xticks([])
plt.xlabel("42 Model Combinations")
plt.ylabel("R² Score")
plt.title("Comparison of All 42 Machine Learning Techniques")

plt.show()
```

```python
import streamlit as st
import pandas as pd
import pickle


# Load model
model = pickle.load(
    open("best_model.pkl", "rb")
)

scaler = pickle.load(
    open("scaler.pkl", "rb")
)

columns = pickle.load(
    open("model_columns.pkl", "rb")
)


st.title("🏠 House Price Prediction System")

st.write(
    "Predict house price using Machine Learning"
)


# User Input

rooms = st.number_input(
    "Number of Rooms",
    min_value=1,
    max_value=10,
    value=3
)


site_area = st.number_input(
    "Site Area (sqm)",
    value=300
)


built_area = st.number_input(
    "Built Area (sqm)",
    value=150
)


years = st.number_input(
    "Property Age (Years)",
    value=5
)


construction = st.selectbox(
    "Construction Materials",
    [
        "Concrete",
        "Mud&Wood"
    ]
)


housing = st.selectbox(
    "Housing Typology",
    [
        "Detached",
        "Semi-detached",
        "Condominium"
    ]
)


land = st.selectbox(
    "Land Value Grade",
    [
        "Low",
        "Medium",
        "High"
    ]
)


road = st.selectbox(
    "Nearest Road Type",
    [
        "Asphalt",
        "Gravel"
    ]
)


CBD = st.number_input(
    "Distance to CBD (km)",
    value=2.0
)


bus = st.number_input(
    "Distance to Bus Station (km)",
    value=1.0
)


school = st.number_input(
    "Distance to School (km)",
    value=1.5
)



if st.button("Predict Price"):


    input_data = pd.DataFrame({

        "Number_of_Rooms":[rooms],
        "Site_Area_sqm":[site_area],
        "Built_Area_sqm":[built_area],
        "Property_Years":[years],

        "Proximity_to_CBD_km":[CBD],
        "Proximity_to_Bus_Station_km":[bus],
        "Proximity_to_Schools_km":[school],

        "Construction_Materials":[construction],
        "Housing_Typology":[housing],
        "Land_Value_Grading":[land],
        "Type_of_Nearest_Road":[road]

    })


    # One hot encoding

    input_data = pd.get_dummies(
        input_data
    )


    # Match training columns

    input_data = input_data.reindex(
        columns=columns,
        fill_value=0
    )


    # Scaling

    input_scaled = scaler.transform(
        input_data
    )


    # Prediction

    price = model.predict(
        input_scaled
    )


    st.success(
        f"Estimated House Price: {price[0]:,.2f} ETB"
    )
```
