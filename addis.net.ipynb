```python
articles = []

for h in soup.find_all('h3', class_='entry-title'):
    a = h.find('a')

    if a:
        articles.append({
            "title": a.get_text(strip=True),
            "link": a['href']
        })
```

```python
print(articles)
```

```python
from bs4 import BeautifulSoup
import requests

url = articles[0]["link"]

res = requests.get(url)
print(res.status_code)

article_soup = BeautifulSoup(res.text, "html.parser")

print(article_soup.title)
```

```python
print(article_soup.prettify()[:5000])
```

```python
for tag in article_soup.find_all(class_=True):
    classes = " ".join(tag.get("class", []))

    if "cat" in classes.lower():
        print(tag)
        print("=" * 50)
```

```python
for a in article_soup.find_all("a", href=True):
    if "/category/" in a["href"]:
        print(a)
```

```python
categories = article_soup.find_all("a", class_="tdb-entry-category")

for cat in categories:
    print(cat.get_text(strip=True))
```

```python
for article in articles:
    res = requests.get(article["link"])
    article_soup = BeautifulSoup(res.text, "html.parser")

    categories = article_soup.find_all("a", class_="tdb-entry-category")

    cat_names = [cat.get_text(strip=True) for cat in categories]

    print("Title:", article["title"])
    print("Link:", article["link"])
    print("Categories:", ", ".join(cat_names))
    print("-" * 80)
```
