```python
#fill missing age
df['age'] = df['age'].fillna(df['age'].median())
df['age'].isnull().sum()
```

```python
df['city'] = df['city'].fillna(df['city'].mode()[0])
df['city'].isnull().sum()
```

```python
df['performance_score'] = df['performance_score'].fillna(df['performance_score'].median())
df['performance_score'].isnull().sum()
```

```python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["salary"] < lower_bound) |
              (df["salary"] > upper_bound)]

print("\nOutliers:")
print(outliers)
```

```python
def age_group(age):
    if age <= 30:
        return "Young"
    elif age <= 45:
        return "Mid"
    else:
        return "Senior"

df['age_group'] = df['age'].apply(age_group)
```

```python
df['age_group'].value_counts()
```

```python
low = df['salary'].quantile(0.33)
high = df['salary'].quantile(0.66)
```

```python
def salary_category(salary):
    if salary <= low:
        return "Low"
    elif salary <= high:
        return "Medium"
    else:
        return "High"

df['salary_category'] = df['salary'].apply(salary_category)
```

```python
df['salary_category'].value_counts()
```

```python
def experience_level(exp):
    if exp <= 3:
        return "Junior"
    elif exp <= 7:
        return "Mid"
    else:
        return "Senior"

df['experience_level'] = df['years_experience'].apply(experience_level)
```

```python
df['experience_level'].value_counts()
```
