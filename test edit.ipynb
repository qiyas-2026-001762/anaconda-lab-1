```python
from bs4 import BeautifulSoup
import requests

res = requests.get("https://addisinsight.net/")

soup = BeautifulSoup(res.text, "html.parser")
```

```python
for h in soup.find_all(['h1','h2','h3']):
    print(h.get_text(strip=True))
```

```python
for a in soup.find_all("a", href=True):
    print(a.get("href"))
```

```python
for h in soup.find_all(['h1', 'h2', 'h3'])[:5]:
    print(h)
    print("=" * 100)
```

```python
for h in soup.find_all('h3', class_='entry-title'):
    a = h.find('a')

    if a:
        title = a.get_text(strip=True)
        link = a['href']

        print(title)
        print(link)
        print()
```

```python
articles = []

for h in soup.find_all('h3', class_='entry-title'):
    a = h.find('a')

    if a:
        articles.append({
            "title": a.get_text(strip=True),
            "link": a['href']
        })
print(articles)
```

```python
from bs4 import BeautifulSoup
import requests

url = articles[0]["link"]

res = requests.get(url)
print(res.status_code)

article_soup = BeautifulSoup(res.text, "html.parser")

print(article_soup.title)
```

```python
print(article_soup.prettify()[:5000])
```

```python
for tag in article_soup.find_all(class_=True):
    classes = " ".join(tag.get("class", []))

    if "cat" in classes.lower():
        print(tag)
        print("=" * 50)
```

```python
for a in article_soup.find_all("a", href=True):
    if "/category/" in a["href"]:
        print(a)
```

```python
categories = article_soup.find_all("a", class_="tdb-entry-category")

for cat in categories:
    print(cat.get_text(strip=True))
```

```python
for article in articles:
    res = requests.get(article["link"])
    article_soup = BeautifulSoup(res.text, "html.parser")

    categories = article_soup.find_all("a", class_="tdb-entry-category")

    cat_names = [cat.get_text(strip=True) for cat in categories]

    print("Title:", article["title"])
    print("Link:", article["link"])
    print("Categories:", ", ".join(cat_names))
    print("-" * 80)
```

```python
for tag in article_soup.find_all(class_=True):
    classes = " ".join(tag.get("class", []))

    if "author" in classes.lower():
        print(tag)
        print("=" * 100)
```

```python
for tag in article_soup.find_all(class_=True):
    classes = " ".join(tag.get("class", []))

    if "date" in classes.lower() or "time" in classes.lower():
        print(tag)
        print("=" * 100)
```

```python
data = []

for h in soup.find_all(['h2','h3']):
    a = h.find("a")

    if a:
        title = a.get_text(strip=True)
        link = a.get("href")

        data.append({
            "title": title,
            "link": link,
            "author": author,
            "date": date
        })

data[:20]
```

```python
import pandas as pd
from dash import Dash, dcc, html, Input, Output, dash_table
import plotly.express as px

# Load your scraped data
df = pd.read_csv("articles.csv")

app = Dash(__name__)

app.layout = html.Div([
    html.H1("Addis Insight Articles Dashboard"),

    html.Label("Select Author"),
    dcc.Dropdown(
        id="author-dropdown",
        options=[{"label": a, "value": a}
                 for a in df["author"].dropna().unique()],
        value=df["author"].iloc[0],
        clearable=False
    ),

    dcc.Graph(id="date-chart"),

    dash_table.DataTable(
        id="article-table",
        columns=[
            {"name": "Title", "id": "title"},
            {"name": "Date", "id": "date"},
            {"name": "Link", "id": "link"},
        ],
        page_size=10,
        style_cell={"textAlign": "left"},
    )
])


@app.callback(
    Output("date-chart", "figure"),
    Output("article-table", "data"),
    Input("author-dropdown", "value")
)
def update_dashboard(selected_author):

    filtered_df = df[df["author"] == selected_author]

    date_counts = (
        filtered_df.groupby("date")
        .size()
        .reset_index(name="articles")
    )

    fig = px.bar(
        date_counts,
        x="date",
        y="articles",
        title=f"Articles by {selected_author}"
    )

    return fig, filtered_df.to_dict("records")


app.run(debug=True)
```
